#  List Slicing



Most important slicing patterns
list[start:end] → extracts elements starting at start up to just before end.

list[start:] → extracts elements from start all the way to the last item.

list[:end] → extracts elements from the first item up to just before end.

list[::-1] → returns the list in reverse order.

```python
fruits = ["Apple", "Banana", "Cherry", "Orange", "Mango", "Pineapple"]
```

```python
print(fruits[1:3])
print(fruits[3:])
print(fruits[:3])
print(fruits[::-1])

# position 2-5
print(fruits[1:4:])
print(fruits[-5:-2])
print(fruits[1:-2])
```

```python
fruits = ["Apple", "Banana", "Cherry", "Orange", "Mango", "Pineapple"]
# position 2-5
print(fruits[3:0:-1])
print(fruits[-3:-6:-1])
print(fruits[-3:0:-1])
```

```python
fruits = ["Apple", "Banana", "Cherry", "Orange", "Mango", "Pineapple"]
fruits[::2]
```

```python
# Most important patterns
fruits = ["Apple", "Banana", "Cherry", "Orange", "Mango", "Pineapple"]
print(fruits[2:5])
print(fruits[3:])
print(fruits[:3])
print(fruits[::-1])
```

# Mutability vs Immutability:
Mutability refers to the ability of an object to be changed or modified after creation.
Mutable data types → can be changed after creation → list, set, dict

Immutable data types → cannot be changed after creation → tuple, int, float, str, bool

```python
old_subscriber = {"rafin@gmail.com", "shafin@gmail.com", "sadia@gmail.com", "hasan@gmail.com"}
new_subscriber = {"sadia@gmail.com", "hasan@gmail.com", "sayem@gmail.com", "adnan@gmail.com"}


# Total unique subscribers
total_unique= old_subscriber.union(new_subscriber)
print("Total union Subscriber:", total_unique)

# Users who shifted to new channel (common subscribers)
shifted_user= old_subscriber.intersection(new_subscriber)
print("user who shifter to new channel:",shifted_user)

# Users who stayed only in old channel
only_old = old_subscriber.difference(new_subscriber)
print("only in old channel",only_old)

# Users who are completely new
only_new= new_subscriber.difference(old_subscriber)
print("only in new channel:", only_new)
```

```python
import timeit

n = 1_000_00
lst = list(range(n))
tpl = tuple(range(n))
st = set(range(n))

list_time= timeit.timeit("9999999 in lst",globals= globals(),number=1000)
tuple_time= timeit.timeit("9999999 in tpl",globals =globals(),number = 1000)
set_time = timeit.timeit("9999999 in st",globals = globals(),number = 1000)

print(f"List : {list_time} ~ {list_time: .02f} seconds")
print(f"Tuples: {tuple_time} ~ {tuple_time: .02f} seconds")
print(f"Set: {set_time} ~ {set_time*100000:.02f} microseconds")
```

Dictionary intro and creation
A dictionary in Python is a built-in, mutable, and ordered collection data type used to store data in key-value pairs.

🌐 Real-World Use Case: Search Engine 🔎

Think of a dictionary like a search engine:

You type a keyword (key)

The search engine returns relevant results (value)

```python
d1 = {"name": "Saihan", "age": 20,"work": "ostad"}

print(d1)
```

```python
d2 = dict(name= "hasan", age= 26,city="Dhaka")
print(d2)
```

```python
d3 = {"a":10,"b":20,"c":30}
d4 = {"a": "abc", "b": "bcd", "c": "cdf"}
d5 = {"a": True, "b": 45, "c": [1,2,3], "d": (7,9,11), "e": {"x": 10, "y": 30}}
d6 = {1: "abc", 2:"bcd", 3:"cdf", 3.14:"pi"}
```

```python
d7 = {
    "name": "hasib",
    "city" : "chittagong",
    "city": "Dhaka"
}

print(d7)

```

Dictionary read/access values
Unlike lists or tuples, dictionaries do not support indexing. You cannot access elements by position; you must use keys to retrieve values.

```python
data = {"name": "Alex", "age": 25}

print(data["name"])   # ✅ key-based access
print(data[0])
```

```python
d = {"name": "Saihan", "age": 20, "work": "Ostad"}
print(d)
```

```python
print(d["name"])
print(d["age"])
print(d["work"])
```

```python
d = {"name": "Saihan", "age": 20, "work": "Ostad", 2: 50, 3: 100}
print(d)
```

```python
data= {"name": "Alex", "age": 25}
print(data["name"])
print(data["age"])
print(data.keys())
print(data.values())
print(data.items())
```

```python
# add item in a dictionary

d= {"name":"saihan","age":20,"work":"ostad"}
print(d)

d["city"]= "dhaka"
print(d)

d.update({"city":"Chittagong"})
print(d)

d.update({"phone":12345})
print(d)
```

```python
# update a item in a dictionary
data= {"name":"Alex","age":20,"work":"ostad"}
print(data)

data["age"] = 26
print(data)

data.update({"age": 30})
print(data)
```

```python
# remove item

d= {'name': 'Saihan', 'age': 20, 'work': 'Ostad', 'city': 'dhaka', 'phone': 12345}
print(d)

d.pop("city")   # remove a key
print(d)

d.popitem()     # remove last key
print(d)

d.clear()
print(d)
```

```python
# A dictionary works like a search engine.
# You give a key, and it instantly returns the value.

student_info= {
    "shorov": {"roll": 123,"city": "dhaka","age": 30},
    "sadia": {"roll": 127,"city": "new york","age": 29}
}

# get full info
print(student_info.get("shorov"))

# get nested value safely
shorov_info = student_info.get("shorov")
print(shorov_info.get("city"))

# Using default value if key does not exist
print(student_info.get("Adeeb","Adeeb not found"))
```
